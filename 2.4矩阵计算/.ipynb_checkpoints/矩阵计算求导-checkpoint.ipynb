```python
# 标量对向量求导 x1 x2  ，标量对列向量求导，得到行向量
# 梯度是上升最快的方向
# 默认向量都是列排列的，因此用 yT表示行向量

# 向量对标量求导，等于各个元素求导组成的向量，仍然是列
# 向量对向量求导，等于一个矩阵，元素是对应yi关于xi的倒数。
# 无论是什么样子的导数，都是遵循复合求导和链式法则
%matplotlib inline
import numpy as np
from matplotlib_inline import backend_inline
import d2l
import torch


def f(x):
    return 3 * x ** 2 - 4 * x
```

```python
def numerical_lim(f, x, h):
    return (f(x + h) - f(x)) / h

h = 1
for i in range(10):
    print(f'h={h:.9f}, numerical limit={numerical_lim(f, 1, h):.9f}')
    h *= -0.1
```
