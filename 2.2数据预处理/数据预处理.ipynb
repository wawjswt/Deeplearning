```python
with open(data_file, 'w') as f:
    f.write('NumRooms, Alley, Price\n') #列title
    f.write('NA,Pave,127500\n')  # 每行表示一个数据样本
    f.write('2,NA,106000\n')
    f.write('4,NA,178100\n')
    f.write('NA,NA,140000\n')
    f.write('6,SB,267000\n')
```

```python
import pandas as pd
data=pd.read_csv(data_file)
data
```

```python
inputs, output = data.iloc[:,0:2], data.iloc[:,-1]
print(inputs), print(output)
inputs
```

```python
inputs = inputs.fillna(inputs.mean(numeric_only=True))
inputs
```

```python
inputs = pd.get_dummies(inputs, dummy_na=True, dtype=int)
inputs  # 进行是否的判断
```

```python
import torch
x,y =torch.tensor(inputs.values), torch.tensor(output.values)
x,y
# 这样就转换成为了一个特征对应一个结果，就可以用于学习模型
```
