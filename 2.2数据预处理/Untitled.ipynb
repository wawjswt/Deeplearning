```python
'''数据预处理是对原始数据进行清洗、转换、编码和规范化，目的是让数据更适合建模。

缺失值是数据中没有记录的空值，常用方法有删除、均值填充、中位数填充、众数填充等。

标准化通常指变成均值为 0、方差为 1；归一化通常把值缩放到某个区间，如 [0, 1]。

类别特征表示类型标签，如颜色、城市；数值特征是可计算的连续数值。

先划分再预处理可以避免数据泄漏。
'''
```
