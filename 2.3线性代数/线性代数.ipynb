```python
A = torch.arange(40, dtype=torch.float32).reshape(2,5,4)
B=A # 不会分配新内存，会同步修改，所以要用clone
B=A.clone()
A, id(A), id(B)
```

```python
# 哈达玛积，矩阵元素对应相乘
A * B
```

```python
A.sum(axis=0) # 第一个维度求和，也就是把两张 5×4 的矩阵叠加起来
```

```python
A.sum(axis=1) # 第二个维度，行求和，列数不变，对应位置行求和
```

```python
A.sum(axis=2) # 第三个维度求和，行数不变，列相加
```

```python
A.sum(axis=[0,1])
```

```python
A.sum(axis=[1,2]) 
```

```python
A.sum(axis=[0,2])
```

```python
A.sum(axis=[0,1,2])
```

```python
A.mean(), A.sum()/ A.numel()  # 均值 = 求和 / 元素个数
```

```python
temp=A.sum(axis=1,keepdims=True)  #保留求和轴的尺寸
temp, temp.shape
```

```python
A/temp
```

```python
x = torch.arange(4,dtype=torch.float32)
y = torch.ones(4,dtype=torch.float32)
x,y,x*y,torch.mul(x,y), torch.dot(x,y)
# 元素对应积，叉乘，点乘
```

```python
# 点乘就是向量元素相乘求和
torch.dot(x,y) == sum(x*y)
```
