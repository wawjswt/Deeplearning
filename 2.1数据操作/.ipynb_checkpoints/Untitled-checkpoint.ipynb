```python
'''
### 三、索引题

1. 给定一个 `3 x 4` 张量，取出第 2 行。
2. 给定一个 `3 x 4` 张量，取出第 3 列。
3. 取出左上角 `2 x 2` 子矩阵。
4. 修改张量中第 1 行第 2 列的元素。
5. 用切片把前两行都改成全 9。'''
```

```python
a=torch.randn(3,4)

a
```

```python
# 3.1
a[1,:]
```

```python
# 3.2
a[:,2]
```

```python
# 3.3
a[:2,:2]
```

```python
# 3.4
a[0,1]=3.1415926
a
```

```python
a[:2,:]=9
a
```
